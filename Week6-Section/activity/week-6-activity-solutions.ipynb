### Codebook

Each row is **one team in one game**, so every game appears twice (once for the home team, once for the away team). The data covers 2022 to 2025 and includes both **regular-season** and **playoff** games.

- `season`: year of the season
- `game_id`, `game_date`: identify the game
- `game_type`: `"regular"` (regular season) or `"playoff"`
- `team`: team name
- `home_away`: `"home"` if the team played in its own arena, `"away"` if not
- `points`: points the team scored in that game
- `opp_points`: points the other team scored in that game
- `win`: 1 if the team won, 0 if it lost
- `reg_win_pct`: the team's win percentage over its regular-season games that season (a measure of how good the team was)
- `assists`, `rebounds`, `turnovers`, `threes_made`: other team stats from the game

**Our question:** does playing at home cause a team to score more points (and win more)?

# Part A: Regular-season games

For Part A, use **regular-season games only** (`game_type == 'regular'`).

**Q1.** Estimate how many more points the home team scores than the away team in **2023 regular-season games**. Save the estimate (home minus away) as `q1.estimate`, its standard error as `q1.se`, and its t-statistic as `q1.t`.

```r
# Step 1: keep only regular-season games from 2023
reg.2023 <- subset(wnba, game_type == 'regular' & season == 2023)

# Step 2: estimate the effect of playing at home (baseline = away)
dim.2023 <- difference_in_means(points ~ home_away, data = reg.2023,
                                condition1 = 'away', condition2 = 'home')
dim.2023

# Step 3: pull out the pieces; t = estimate / standard error
q1.estimate <- dim.2023$coefficients
q1.se <- dim.2023$std.error
q1.t <- q1.estimate / q1.se
```

### Helper functions (given)

Just run this cell. `re.randomize()` shuffles the home/away labels in a dataset, and `compute.home.effect()` returns the home-minus-away gap in points for a dataset. They work like the functions from lecture.

```r
re.randomize <- function(input.data) {
    input.data$home_away <- sample(input.data$home_away)
    return(input.data)
}

compute.home.effect <- function(input.data) {
    result <- difference_in_means(points ~ home_away, data = input.data,
                                  condition1 = 'away', condition2 = 'home')
    return(result$coefficients)
}
```

**Q4.** Suppose the 2023 regular season had been **4 times as long**: 4 times as many home and away team-games, with the same standard deviations of points. What standard error would you expect for the home-minus-away gap? Save it as `q5.se.4x`.

```r
# Step 1: standard deviation and number of games in each group
home.2023 <- subset(reg.2023, home_away == 'home')
away.2023 <- subset(reg.2023, home_away == 'away')
s.home <- sd(home.2023$points)
s.away <- sd(away.2023$points)
n.home <- 241   # from table(reg.2023$home_away)
n.away <- 241

# Step 2: SE formula with 4x the sample in each group
q5.se.4x <- (s.home^2 / (4 * n.home) + s.away^2 / (4 * n.away))^0.5

# Shortcut: 4x the data cuts the SE in half, so q1.se / 2 gives the same answer
```

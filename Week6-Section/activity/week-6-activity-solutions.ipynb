### Codebook

### Helper functions (given)

Just run this cell. `re.randomize()` shuffles the home/away labels in a dataset, and `compute.home.effect()` returns the home-minus-away gap in points for a dataset. They work like the functions from lecture.

```r
re.randomize <- function(input.data) {
    input.data$home_away <- sample(input.data$home_away)
    return(input.data)
}

compute.home.effect <- function(input.data) {
    result <- difference_in_means(points ~ home_away, data = input.data,
                                  condition1 = 'away', condition2 = 'home')
    return(result$coefficients)
}
```

---
Some of this is from the lecture from Monday so it may not make sense right now. But I thought it might be helpful to let you see it right now so you are learning it over a longer period of time. 

Imagine someone who says home/away makes no difference. The code below shuffles the home/away labels in the **2023 regular-season games 1,000 times**, recomputing the home-minus-away gap each time and saves the 1,000 results as `q3.sims`.

(This cell builds its own copy of the 2023 data, so it works no matter what you named your datasets.)

```r
set.seed(3)
shuffle.data <- subset(wnba, game_type == 'regular' & season == 2023)
q3.sims <- replicate(1000, compute.home.effect(re.randomize(shuffle.data)))
```

---
We can then graph the estimates and see how rare the results we got are. The red line is the actual 2023 estimate.

```r
qplot(q3.sims, binwidth = .25) + geom_vline(xintercept = compute.home.effect(shuffle.data), color = 'red')
```

```r
sd(q3.sims)
```

**Q4.** Suppose the 2023 regular season had been **4 times as long**: 4 times as many home and away team-games, with the same standard deviations of points. What standard error would you expect for the home-minus-away gap? Save it as `q4.se.4x`.

```r
# Option 1: use the SE formula with 4x the sample in each group
home.2023 <- subset(reg.2023, home_away == 'home')
away.2023 <- subset(reg.2023, home_away == 'away')
s.home <- sd(home.2023$points)
s.away <- sd(away.2023$points)
n.home <- 241   # from table(reg.2023$home_away)
n.away <- 241
q4.se.4x <- (s.home^2 / (4 * n.home) + s.away^2 / (4 * n.away))^0.5

# Option 2 (shortcut): 4x the data cuts the SE in half, so take the Std. Error from the Q1 output and divide by 2
q4.se.4x
```

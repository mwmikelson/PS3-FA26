## Week 12 Section Activity Solutions: Analyzing Experiments with Regression and Reading Regression Tables

First, let's read in the data.

```r
#import libraries
library(testthat)

data <- read.csv("ps3-misinformation-india.csv")
head(data)
```

The data comes from a 2025 study by Amar, Badrinathan, Chauchard, and Sichart of a classroom program meant to counter misinformation in villages in Bihar, India. Villages were randomly assigned to either a media literacy class (the treatment) or a spoken English class (the control), and each class met four times over several months. Each row is one student. After the classes, students answered survey questions that the authors combined into seven outcomes; students who did not complete this survey have `NA` for these outcomes.

Each outcome is scaled so that the control group has a mean of 0 and a standard deviation of 1, and higher values are better. So an effect of 0.3 means 0.3 control-group standard deviations. We use simpler regressions than the authors did, so our numbers will be close to, but not exactly the same as, the numbers in the paper.

Here is a quick rundown of what each column means:

- `treatment`: Whether the student's village was assigned to the media literacy class (`"Media Literacy"`) or the spoken English class (`"Control (Spoken English)"`)
- `accuracy_discernment`: How well the student told true health stories from false ones when rating how accurate they were
- `sharing_discernment`: How much more the student said they would share true stories than false ones
- `health_preferences`: Interest in health news, views on vaccine safety, and (lack of) reliance on non-scientific remedies
- `source_discernment`: How much the student trusted reliable sources of health information relative to unreliable ones
- `engagement_attitude`: Attitudes toward fact-checking and correcting misinformation
- `engagement_behavior`: Willingness to take costly actions against misinformation, like becoming a "truth ambassador"
- `awareness`: Awareness of the threat of misinformation and of biases in the news
- `age`: Age of the student (measured before the classes)
- `girl`: Whether the student is a girl (1 = girl, 0 = boy; measured before the classes)
- `grade`: Grade in school (measured before the classes)
- `asset_index`: Index of the household's assets, where higher values mean a wealthier household and the average student is 0 (measured before the classes)
- `mobile_internet`: Whether the household has mobile internet access (1 = yes, 0 = no; measured before the classes)
- `reading`: Reading skill score from 0 to 2 (measured before the classes)
- `science`: Science knowledge score from 0 to 8 (measured before the classes)
- `sessions_attended`: Number of the four classes the student attended (0 to 4)

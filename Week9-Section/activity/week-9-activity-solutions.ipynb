Sometimes we want the proportion of people who do two things at once, like supporting two different policies. The trick is to make a new column that equals `1` only for those people.

Multiplying two `0`/`1` columns does exactly this: `1 * 1 = 1`, but any combination that includes a `0` (`1 * 0`, `0 * 1`, or `0 * 0`) gives `0`. So the new column is `1` for respondents who support **both** policies and `0` for everyone else.

And because the new column is just `0`s and `1`s, its weighted mean is the weighted proportion of people with a `1`. That is the weighted proportion who support both policies. Run the cell below to see this in action for `legal_status` and `border_patrol`.

```r
# This cell re-reads the data, so it works no matter what you did above.
demo.data <- read.csv("ps3_cces2020_week9.csv")

# Step 1: Make a new column that is 1 only if the person supports both policies
demo.data$both <- demo.data$legal_status * demo.data$border_patrol

# Look at each different combination of the two policies next to the new column
# (unique() just shows each different row once)
unique(data.frame(legal_status = demo.data$legal_status,
                  border_patrol = demo.data$border_patrol,
                  both = demo.data$both))

# Step 2: Take the weighted mean of the new column
weighted.mean(demo.data$both, demo.data$vvweight)
```

# Part B: Practice with `difference_in_means()`

This part contains practice estimating differences between groups using the CES data.

```r
# Run this cell first for Part B
# import libraries
library(testthat)
library(ottr)
library(estimatr)

# Read in data
data <- read.csv("ps3_cces2020_week9.csv")
head(data)
```

This part uses the same CES 2020 data as the other parts. Here are the variables used in this part:

- `tv_news`: Whether the person watched TV news in the past 24 hours (`1` = yes, `0` = no)
- `social_media`: Whether the person used social media in the past 24 hours (`1` = yes, `0` = no)
- `party`: The person's party: `"Democrat"`, `"Republican"`, `"Independent"`, or `"Other"`
- `medicare_expand`: Whether the person supports expanding Medicare to a single comprehensive public health care coverage program that would cover all Americans (`1` = yes, `0` = no)
- `repeal_aca`: Whether the person supports repealing the entire Affordable Care Act (`1` = yes, `0` = no)
- `restore_mandate`: Whether the person supports restoring the Affordable Care Act's mandate that all individuals be required to purchase health insurance (`1` = yes, `0` = no)

**Note:** This is survey data, not an experiment: nobody was randomly assigned to watch TV news, use social media, or listen to the radio. So we use "treatment group" and "control group" only as labels for the two groups being compared (the treatment group is the group of interest and the control group is the comparison group). A difference between the groups is not necessarily a causal effect.

**Question 8.** Estimate the difference in support for expanding Medicare between people who watched TV news and people who did not.

- **Treatment group:** Respondents who watched TV news in the past 24 hours (`tv_news = 1`)

Save the full `difference_in_means()` output as `tv.dim`.

```r
# The treatment is watching TV news; the control is not watching it.
# Control goes in condition1 and treatment goes in condition2.
tv.dim <- difference_in_means(medicare_expand ~ tv_news, data = data,
                              condition1 = 0, condition2 = 1)
tv.dim
```

**Question 24.** A researcher surveys 200 people and finds that people who listened to radio news were 4 percentage points more supportive of a policy than people who did not, with a p-value of 0.40. She concludes: "There is no difference between these groups in the population." What is the best response?

- `'a'`: She is right. A large p-value proves that the null hypothesis is true.
- `'b'`: She can't conclude that. A large p-value means the survey didn't provide strong evidence of a difference, but a survey this small might just be too noisy to detect a real difference.
- `'c'`: She is wrong, because any difference at all would have given a p-value below 0.05.
- `'d'`: A p-value of 0.40 means there is a 40% chance that the null hypothesis is true, so she is probably right.

Put your answer in the quotes below:

```r
q24.answer <- 'b'
```

```r
. = ottr::check("tests/q24.R")
```

**Solution:** `'b'`. Failing to reject the null hypothesis is not the same as showing the null hypothesis is true. A real difference can easily go undetected in a small survey.

Sometimes we want the proportion of people who do two things at once, like supporting two different policies. The trick is to make a new column that equals `1` only for those people.

Multiplying two `0`/`1` columns does exactly this: `1 * 1 = 1`, but any combination that includes a `0` (`1 * 0`, `0 * 1`, or `0 * 0`) gives `0`. So the new column is `1` for respondents who support **both** policies and `0` for everyone else.

And because the new column is just `0`s and `1`s, its weighted mean is the weighted proportion of people with a `1`. That is the weighted proportion who support both policies. Run the cell below to see this in action for `legal_status` and `border_patrol`.

````r
# This cell re-reads the data, so it works no matter what you did above.
demo.data <- read.csv("ps3_cces2020_immigration.csv")

# Step 1: Make a new column that is 1 only if the person supports both policies
demo.data$both <- demo.data$legal_status * demo.data$border_patrol

# Look at each different combination of the two policies next to the new column
# (unique() just shows each different row once)
unique(data.frame(legal_status = demo.data$legal_status,
                  border_patrol = demo.data$border_patrol,
                  both = demo.data$both))

# Step 2: Take the weighted mean of the new column
weighted.mean(demo.data$both, demo.data$vvweight)
````

# Part B: Practice with `difference_in_means()`

This part contains practice estimating treatment effects from three real randomized experiments.

````r
# Run this cell first for Part B!
# import libraries
library(testthat)
library(ottr)
library(estimatr)

# Read in data
resume <- read.csv("ps3_resume.csv")
social <- read.csv("ps3_social_pressure.csv")
star <- read.csv("ps3_star.csv")
head(resume)
head(social)
head(star)
````

**Reminder: how `difference_in_means()` works.**

```r
difference_in_means(outcome ~ treatment.column, data = my.data,
                    condition1 = "control group", condition2 = "treatment group")
```

- The estimate it reports is the average outcome in `condition2` **minus** the average outcome in `condition1`.
- So, to estimate a treatment effect, put the **control** group in `condition1` and the **treatment** group in `condition2`. If you swap them, the sign of your estimate flips.
- If the treatment column has more than two groups, `condition1` and `condition2` pick which two groups to compare.
- Text values go in quotes (like `"Control"`); numbers don't (like `2`).

Here's what the variables mean in the three datasets.

**`resume`**: A field experiment where researchers sent fake resumes to real job ads, randomly assigning each resume a first name. Each row is a resume.

- `firstname`: The first name on the resume
- `sex`: Whether the name is female or male (`"female"` or `"male"`)
- `race`: Whether the first name sounds white or Black (`"white"` or `"black"`). Names were randomly assigned.
- `call`: Whether the resume got a callback from the employer (`1` = yes, `0` = no)

**`social`**: A field experiment where registered voters were randomly assigned to get a mailer before the 2006 primary election, or no mailer at all. Each row is a voter.

- `messages`: What the voter was sent: `"Control"` (no mailer), `"Civic Duty"` (a mailer reminding them that voting is a civic duty), `"Hawthorne"` (a mailer saying they are being studied), or `"Neighbors"` (a mailer saying their neighbors will be able to see whether they voted)
- `primary2006`: Whether the voter voted in the 2006 primary (`1` = yes, `0` = no)
- `primary2004`: Whether the voter voted in the 2004 primary, _before_ the experiment (`1` = yes, `0` = no)
- `sex`: `"female"` or `"male"`
- `yearofbirth`, `hhsize`: Year of birth and household size

**`star`**: A field experiment where Tennessee students were randomly assigned to different kinds of classrooms starting in kindergarten. Each row is a student.

- `classtype`: The type of classroom: `1` = small class, `2` = regular-size class, `3` = regular-size class with a teacher's aide
- `g4reading`, `g4math`: The student's reading and math test scores in 4th grade. Some students are missing scores (`NA`); `difference_in_means()` just skips these, so you don't need to do anything about them.
- `hsgrad`: Whether the student graduated high school (`1` = yes, `0` = no)
- `race`: Student's race

# Part C: Heterogeneous treatment effects

In this part, you'll look at whether the effect of a treatment is the same for everyone.

````r
# Run this cell first for Part C!
# import libraries
library(testthat)
library(ottr)
library(estimatr)

# Read in data
social <- read.csv("ps3_social_pressure.csv")
head(social)
````

**What are heterogeneous treatment effects?** A treatment effect can be different for different kinds of people. The average treatment effect (ATE) for everyone can hide this: an average effect of +5 might be +10 for one group and 0 for another.

To find out whether effects differ across groups, we estimate the treatment effect _separately within each group_, by running the same difference in means in each subset of the data. If the estimates are very different (taking into account their uncertainty), that is evidence of heterogeneous treatment effects.

You'll use the same data as in Part B, but you don't need to have done Part B. The data are from a field experiment where registered voters were randomly assigned to get a mailer before the 2006 primary election, or no mailer at all. Each row is a voter.

- `messages`: What the voter was sent: `"Control"` (no mailer), `"Civic Duty"`, `"Hawthorne"`, or `"Neighbors"` (a mailer saying their neighbors will be able to see whether they voted)
- `primary2006`: Whether the voter voted in the 2006 primary (`1` = yes, `0` = no)
- `primary2004`: Whether the voter voted in the 2004 primary, _before_ the experiment (`1` = yes, `0` = no)
- `sex`: `"female"` or `"male"`
- `yearofbirth`, `hhsize`: Year of birth and household size

In every question in this part:

- **Treatment group:** Voters sent the "Neighbors" mailer (`messages` is `"Neighbors"`)
- **Control group:** Voters sent no mailer (`messages` is `"Control"`)
- **Outcome:** `primary2006`

**Reminder: how `difference_in_means()` works.**

```r
difference_in_means(outcome ~ treatment.column, data = my.data,
                    condition1 = "control group", condition2 = "treatment group")
```

- The estimate it reports is the average outcome in `condition2` **minus** the average outcome in `condition1`.
- So, to estimate a treatment effect, put the **control** group in `condition1` and the **treatment** group in `condition2`. If you swap them, the sign of your estimate flips.
- If the treatment column has more than two groups, `condition1` and `condition2` pick which two groups to compare.
- Text values go in quotes (like `"Control"`); numbers don't (like `2`).

**A quick reminder about p-values.** A p-value starts from a _null hypothesis_, which says that the treatment has no effect. The p-value tells us how likely it would be to see an estimate at least as far from zero as ours _if the null hypothesis were true_ and only chance (from random assignment) were at work. A small p-value means our estimate would be surprising if there were really no effect. By convention, a p-value below 0.05 is called "statistically significant."

When you run `difference_in_means()`, the p-value is in the column `Pr(>|t|)` of the output.

You'll use two datasets. You don't need to have done the other parts.

**`resume`**: A field experiment where researchers sent fake resumes to real job ads, randomly assigning each resume a first name. Each row is a resume.

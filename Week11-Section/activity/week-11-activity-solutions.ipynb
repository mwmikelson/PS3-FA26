## Week 11 Section Activity Solutions: Bivariate and Multivariate Regression

Solutions for the Week 11 section activity.

First, let's read in the data.

```r
#import libraries
library(testthat)
library(ggplot2)

data <- read.csv("ps3-ww1-nationalist-vote.csv")
head(data)
```

The data come from a 2023 study by De Juan, Haass, Koos, Riaz, and Tichelbaecker of whether WW1 battle deaths increased support for right-wing nationalist parties, including the Nazi Party. Each row is a county in Germany, and the election is the November 1932 Reichstag election. Counties missing any of these variables are dropped, so every regression uses the same counties.

Here is a quick rundown of what each column means:

- `county`: County name
- `nationalist_vote_percent`: Combined vote share of the two main right-wing nationalist parties, the Nazi Party and the German Nationalist People's Party, in the November 1932 election (percent)
- `fatality_percent`: Percent of the county's WW1 soldier casualties (soldiers who died or were wounded) who died; this is the county's casualty fatality rate (percent)
- `lutheran_percent`: Percent of the county's population that was Lutheran
- `city`: Whether the county is a city (1 = city, 0 = not a city)
- `prewar_conservative_vote_percent`: Vote share of conservative parties in the county in elections before WW1 (percent)

## Bivariate regression

How does the casualty fatality rate relates to percent of the vote going to nationalist parties?

-----

**Question 1.** Using bivariate regression, estimate the relationship between the nationalist vote share (`nationalist_vote_percent`, as the outcome) and the casualty fatality rate (`fatality_percent`, as the predictor variable). Use all counties in `data`. Save the summary of the regression as `q1`.

```r
q1 <- summary(lm(nationalist_vote_percent ~ fatality_percent, data))
q1
```

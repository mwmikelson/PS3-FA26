## Estimated Regression Equations

A regression table gives you the estimates, and the estimated regression equation turns those estimates into a formula that predicts the outcome. The estimated regression equation has one term for the intercept and one term for each explanatory variable:

$$\widehat{\text{outcome}} = \text{intercept} + (\text{estimate on } x_1) \times x_1 + (\text{estimate on } x_2) \times x_2$$

The hat on the outcome means it is a predicted value: the average outcome we expect for an observation with those values of the explanatory variables, not an actual value from the data. The intercept is the predicted outcome when every explanatory variable equals 0. To predict the outcome for a hypothetical observation, replace each $x$ with that observation's value and calculate. A regression with one explanatory variable has only the intercept term and one estimate term.

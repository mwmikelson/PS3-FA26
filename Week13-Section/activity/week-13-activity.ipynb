## Election Observers and MP Spending in Ghana

In this section, we will read regression output from an experiment on election observers in Ghana.

Ofosu (2019) uses a field experiment from Ghana's 2012 election. Researchers randomly assigned how intensely election observers were deployed in each constituency. The treatment group is constituencies with medium or high intensity election observers (`observers` = 1) and the control group is constituencies with low intensity election observers (`observers` = 0). The data has 60 constituencies, 47 in the treatment group and 13 in the control group.

After the election, Ofosu looked at how the winning MPs behaved. MPs control a Constituency Development Fund (CDF) that they spend on local projects. Each row of the data is one constituency, with these variables:

- `constituency`: name of the constituency
- `region`: region of Ghana
- `observers`: 1 if the constituency had medium or high intensity election observers (treatment), 0 if it had low intensity election observers (control)
- `public_pct`: percent of the constituency's CDF funds that the MP spent on public goods between 2014 and 2016
- `absence_pct`: the MP's absence rate in parliament, in percent
- `incumbent_npp`: 1 if the constituency's MP is from the NPP, 0 if the MP is from the NDC (Ghana's two main parties); this was not randomly assigned

The data uses survey experiments from Clayton et al. on electoral gender quotas in 12 democracies. Each respondent read a short news article about a local council that decided to require workplace training for certain employees. Then they rated how fair and legitimate the council's process and decision were.

```r
library(estimatr)
library(ottr)
clayton <- read.csv("clayton-clean.csv")
head(clayton)
```

| Variable | Meaning |
|---|---|
| `treat` | Council in the article: `'all_male'` (8 men), `'balanced'` (4 men and 4 women), or `'quota'` (4 men and 4 women, elected under a rule requiring parties to run equal numbers of male and female candidates) |
| `issue` | What the council's training requirement was about: `'harassment'` (sexual harassment) or `'animals'` (animal mistreatment on farms). The animals version was not used in Australia, France, or Norway. |
| `country` | Respondent's country |
| `procedural` | How fair the **decision-making process** seemed (higher = more legitimate), in standard deviations from the country average |
| `substantive` | How fair the **decision itself** seemed (higher = more legitimate), in standard deviations from the country average |

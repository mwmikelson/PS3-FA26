## Part 1: What is a function?

A function takes some **input**, does something with it, and gives back an **output**. For example, in `mean(states$murder)` the input is a column of numbers and the output is one number.

Why write your own? Whenever you find yourself copying and pasting the same code and only changing a number or a name, you can write a function once and reuse it as many times as you want.

Here is a function that doubles a number. Run the cell below. Nothing will print, because all you did was *save* the function, just like saving a variable.

````r
double_it <- function(x) {
  return(x * 2)
}
````

Here is what each part means:

- `double_it` is the **name** of the function. You pick the name, just like when you save a variable.
- `function(x)` says that this function takes one input. Inside the function, that input is called `x`. It is a placeholder for whatever you give the function later.
- Everything between `{` and `}` is the **body**, the code that runs when you use the function.
- `return(...)` says what the function should give back as its output.

Now let's use it. Using a function is called *calling* it:

````r
double_it(10)
double_it(head(states$murder))
````

A function can take more than one input. Separate them with commas. When you call the function, R matches your inputs to the placeholders in order, or you can name them yourself with `=` and give them in any order:

````r
add_them <- function(a, b) {
  return(a + b)
}

add_them(3, 4)
add_them(b = 10, a = 1)
````

## Part 2: Functions that write text

The function `paste0()` glues pieces of text and numbers together into one piece of text. Each piece is separated by a comma, and text goes inside quotes:

````r
paste0("Hello, ", "world")
paste0("I am ", 20, " years old")
````

Notice that `paste0()` does **not** add any spaces for you. If you want a space, you have to put it inside the quotes yourself (like the space after the comma in `"Hello, "`).

You can also put calculations inside `paste0()`. The function `round(x, 1)` rounds `x` to 1 decimal place (use `2` for two decimal places, and so on). Numbers can be messy, so rounding makes text easier to read:

````r
mean(states$murder)
round(mean(states$murder), 1)
paste0("The average murder rate is ", round(mean(states$murder), 1), " per 100,000 people.")
````

Now let's put `paste0()` inside a function. This function takes a name and a city, and returns a greeting:

````r
greet <- function(name, city) {
  return(paste0("Hi ", name, ", welcome to ", city, "!"))
}

greet("Sam", "Berkeley")
````

## Part 3: Functions that take a whole dataset

A function's input does not have to be a single number. It can be a whole data frame! Here is a function that takes a data frame and reports the average land area of the states in it. Inside the function, `data` is a placeholder for whatever data frame you hand in:

````r
average_area <- function(data) {
  return(paste0("Average area: ", round(mean(data$area), 1), " square miles"))
}

average_area(states)
average_area(subset(states, region == "West"))
````

The same function gave two different answers because we gave it two different data frames. To make this work, the function assumes the data frame it is given has a column called `area`.

One more tool: `c()` combines several things into one output. We can use it to make a function that returns several lines of text at once. You can put `paste0()` calls inside `c()`, separated by commas.

````r
c("first line", "second line", "third line")
````

## Part 4: Functions that take a value like `"South"`

Inputs can also be values you want to filter on, like the name of a region. Here is a function that takes a data frame and a region name and returns the average income in that region:

````r
mean_income_in <- function(data, region_name) {
  region_data <- subset(data, region == region_name)
  return(mean(region_data$income))
}

mean_income_in(states, "South")
mean_income_in(states, "West")
````

Two things to notice. First, the input is the *value* `"South"`, not a column name. Second, `region_data` was created *inside* the function, so it disappears as soon as the function finishes. Only what you `return()` comes out.

(We called the input `region_name` rather than `region` on purpose. `region` is already the name of a column in `states`, and R gets confused if an input has the same name as a column you are using inside `subset()`.)

You can also put `difference_in_means()` inside a function.
